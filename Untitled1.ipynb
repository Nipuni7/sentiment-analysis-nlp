### 01 — Dataset Load

```python
import os
import zipfile
import pandas as pd

# 1. Paths define කිරීම
zip_path = r"C:\Users\ASUS\Desktop\ml train\sentiment+labelled+sentences.zip"
extract_dir = r"C:\Users\ASUS\Desktop\ml train\extracted_sentiment"

# 2. ZIP file එක extract කිරීම (කලින් extract කරලා තිබුණත් override වේ)
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_dir)

base_path = os.path.join(extract_dir, "sentiment labelled sentences")

# 3. Files 3ම (Yelp, Amazon, IMDB) load කර එකතු කිරීම
files = ['yelp_labelled.txt', 'amazon_cells_labelled.txt', 'imdb_labelled.txt']
dfs = []

for file in files:
    file_path = os.path.join(base_path, file)
    temp_df = pd.read_csv(file_path, sep='\t', header=None, names=['text', 'sentiment'])
    dfs.append(temp_df)

df = pd.concat(dfs, ignore_index=True)

# 4. Data inspect කිරීම
print("Dataset Shape:", df.shape)
print("\nClass Distribution:")
print(df['sentiment'].value_counts())
df.head()
```

### 02 — Text Preprocessing

```python
import re
import nltk
from nltk.corpus import stopwords


nltk.download('stopwords')


stop_words = set(stopwords.words('english'))
negation_words = {'not', 'no', 'never', 'nor', 'neither'}
custom_stop_words = stop_words - negation_words

def preprocess_text(text):
    # 1. Lowercase
    text = text.lower()
    # 2. Remove punctuation, symbols & numbers
    text = re.sub(r'[^a-z\s]', '', text)
    # 3. Remove extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()
    # 4. Remove stop words (preserving negations)
    words = [w for w in text.split() if w not in custom_stop_words]
    return ' '.join(words)


df['clean_text'] = df['text'].apply(preprocess_text)

df[['text', 'clean_text', 'sentiment']].head()
```

### 03: Split the Dataset

```python
from sklearn.model_selection import train_test_split

# Features (cleaned text) and target (sentiment labels)
X = df['clean_text']
y = df['sentiment']

# 80% Train, 20% Test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print(f"Total samples   : {len(df)}")
print(f"Training samples: {len(X_train)}")
print(f"Testing samples : {len(X_test)}")
```

### 4 & 5: Tokenization and Sequence Padding

```python
!pip install tensorflow
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
import numpy as np

# 1. Define hyperparameters
vocab_size = 5000       # Maximum vocabulary size
max_length = 50         # Maximum sequence length per review
embedding_dim = 64      # Dimension of word embedding vectors
oov_tok = "<OOV>"       # Out-Of-Vocabulary token for unseen words

# 2. Fit Tokenizer ONLY on training texts
tokenizer = Tokenizer(num_words=vocab_size, oov_token=oov_tok)
tokenizer.fit_on_texts(X_train)

# 3. Convert texts to integer sequences
X_train_seq = tokenizer.texts_to_sequences(X_train)
X_test_seq = tokenizer.texts_to_sequences(X_test)

# 4. Pad sequences to ensure uniform input length
X_train_pad = pad_sequences(X_train_seq, maxlen=max_length, padding='post', truncating='post')
X_test_pad = pad_sequences(X_test_seq, maxlen=max_length, padding='post', truncating='post')

# Convert targets to numpy arrays
y_train_arr = np.array(y_train)
y_test_arr = np.array(y_test)

print(f"Padded Training shape : {X_train_pad.shape}")
print(f"Padded Testing shape  : {X_test_pad.shape}")
```

### 6: Build and Train the Neural Network

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, GlobalAveragePooling1D, Dense, Dropout

# 1. Build Neural Network Architecture
model = Sequential([
    # Embedding layer maps integer word indices to dense 64-dimensional vectors
    Embedding(input_dim=vocab_size, output_dim=embedding_dim, input_length=max_length),
    
    # Pools sentence representations by averaging across token vectors
    GlobalAveragePooling1D(),
    
    # Fully connected hidden layer
    Dense(32, activation='relu'),
    Dropout(0.3),  # Helps prevent overfitting
    
    # Output layer for binary classification (0 = Negative, 1 = Positive)
    Dense(1, activation='sigmoid')
])

# 2. Compile model
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()

# 3. Train model
history = model.fit(
    X_train_pad, y_train_arr,
    epochs=15,
    batch_size=32,
    validation_data=(X_test_pad, y_test_arr),
    verbose=1
)
```

### 07: Plot Training & Validation Accuracy / Loss

```python
import matplotlib.pyplot as plt

# 1. Plot Accuracy
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Train Accuracy', color='blue')
plt.plot(history.history['val_accuracy'], label='Val Accuracy', color='orange')
plt.title('Model Accuracy over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

# 2. Plot Loss
plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Train Loss', color='blue')
plt.plot(history.history['val_loss'], label='Val Loss', color='orange')
plt.title('Model Loss over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

# Print final test evaluation metrics
test_loss, test_acc = model.evaluate(X_test_pad, y_test_arr, verbose=0)
print(f"Test Accuracy: {test_acc * 100:.2f}%")
print(f"Test Loss    : {test_loss:.4f}")
```

### 08: Predict Sentiment on Brand New Sentences

```python
def predict_nn_sentiment(sentence):
    # 1. Clean using your preprocessing function
    cleaned = preprocess_text(sentence)
    # 2. Convert to integer sequences using the fitted tokenizer
    seq = tokenizer.texts_to_sequences([cleaned])
    # 3. Pad to uniform length (50)
    padded = pad_sequences(seq, maxlen=max_length, padding='post', truncating='post')
    # 4. Predict probability using the Neural Network
    prob = model.predict(padded, verbose=0)[0][0]
    
    label = "Positive" if prob >= 0.5 else "Negative"
    confidence = prob if prob >= 0.5 else (1 - prob)
    return label, confidence * 100

# Sample reviews to test
sample_reviews = [
    "The camera and battery life are simply outstanding!",
    "Terrible customer service and the item broke on the first day.",
    "The plot was rather slow, but the cinematography was breathtaking.",
    "Not worth the price, highly disappointed."
]

print("--- Neural Network Predictions ---")
for review in sample_reviews:
    pred_label, conf = predict_nn_sentiment(review)
    print(f"Review     : \"{review}\"")
    print(f"Prediction : {pred_label} ({conf:.2f}% confidence)")
    print("-" * 55)
```

### 09: Final Evaluation Summary & Comparison

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, f1_score

# 1. Vectorize text with Bag-of-Words
bow_vectorizer = CountVectorizer(stop_words='english')
X_train_bow = bow_vectorizer.fit_transform(X_train)
X_test_bow = bow_vectorizer.transform(X_test)

# 2. Train Logistic Regression
lr_model = LogisticRegression(max_iter=1000, random_state=42)
lr_model.fit(X_train_bow, y_train)
lr_pred = lr_model.predict(X_test_bow)

# 3. Train Linear SVM
svm_model = LinearSVC(random_state=42)
svm_model.fit(X_train_bow, y_train)
svm_pred = svm_model.predict(X_test_bow)

# 4. Neural Network predictions on test set
nn_probs = model.predict(X_test_pad, verbose=0)
nn_preds = (nn_probs >= 0.5).astype(int).flatten()

# 5. Compile and display comparison table
comparison_data = {
    "Model": [
        "Logistic Regression (BOW)",
        "Linear SVM (BOW)",
        "Sequential Neural Network (Embedding)"
    ],
    "Accuracy": [
        accuracy_score(y_test, lr_pred),
        accuracy_score(y_test, svm_pred),
        accuracy_score(y_test_arr, nn_preds)
    ],
    "F1-Score": [
        f1_score(y_test, lr_pred),
        f1_score(y_test, svm_pred),
        f1_score(y_test_arr, nn_preds)
    ]
}

comparison_df = pd.DataFrame(comparison_data)
print("=== Model Performance Comparison ===")
comparison_df
```

### Train the LSTM Model (Board Architecture)

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout

# Whiteboard Architecture: Embedding -> LSTM -> Dense -> Output
lstm_model = Sequential([
    # 1. Embedding Layer
    Embedding(input_dim=vocab_size, output_dim=embedding_dim, input_length=max_length),
    
    # 2. LSTM Layer (Captures word order and sequence context)
    LSTM(64, dropout=0.2, recurrent_dropout=0.2),
    
    # 3. Dense Hidden Layer
    Dense(32, activation='relu'),
    Dropout(0.3),
    
    # 4. Dense Output Layer (Binary Classification)
    Dense(1, activation='sigmoid')
])

lstm_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

lstm_model.summary()

# Train the LSTM model
lstm_history = lstm_model.fit(
    X_train_pad, y_train_arr,
    epochs=10,
    batch_size=32,
    validation_data=(X_test_pad, y_test_arr),
    verbose=1
)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Bidirectional, LSTM, Dense, Dropout

# Clear previous session
tf.keras.backend.clear_session()

# Build Model
lstm_model = Sequential([
    # Explicit input shape for Keras 3
    tf.keras.Input(shape=(max_length,)),
    Embedding(input_dim=vocab_size, output_dim=embedding_dim),
    
    # Bidirectional LSTM to read sentences forward and backward
    Bidirectional(LSTM(32, dropout=0.1)),
    
    Dense(32, activation='relu'),
    Dropout(0.2),
    Dense(1, activation='sigmoid')
])

# Adam optimizer with learning rate
optimizer = tf.keras.optimizers.Adam(learning_rate=0.001)

lstm_model.compile(
    optimizer=optimizer,
    loss='binary_crossentropy',
    metrics=['accuracy']
)

lstm_model.summary()

# Train model
lstm_history = lstm_model.fit(
    X_train_pad, y_train_arr,
    epochs=6,
    batch_size=32,
    validation_data=(X_test_pad, y_test_arr),
    verbose=1
)
```

```python
# Evaluate the Bi-LSTM model on the test set
lstm_loss, lstm_acc = lstm_model.evaluate(X_test_pad, y_test_arr, verbose=0)
print(f"Bi-LSTM Test Accuracy: {lstm_acc * 100:.2f}%")
print(f"Bi-LSTM Test Loss    : {lstm_loss:.4f}")
```
